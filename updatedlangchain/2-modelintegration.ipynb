### Streaming
Most Models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as the are produced. You can use a loop to process each chunk in real-time

```python
response = ollama_model.invoke(
    "Write me a 200-word paragraph on Artificial Intelligence"
)
response.content
```

```python
ollama_model.stream("Write me a 200-word paragraph on Artificial Intelligence")
```

```python
for chunk in ollama_model.stream(
    "Why do parrots have colorful feathers?"
):
    print(chunk.text, end="", flush=True)
```

### Batch

Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel

```python
responses = ollama_model.batch([
    "Why do parrots have colorful feathers",
    "How do airplanes fly?",
    "What is quantum computing?"
])
for response in responses:
    print(response)
```

```python
responses = ollama_model.batch([
    "Why do parrots have colorful feathers",
    "How do airplanes fly?",
    "What is quantum computing?"
],
    config = {
        "max_concurrency": 5, # Limit to 5 parallel calls
    }
)
for response in responses:
    print(response)
```
